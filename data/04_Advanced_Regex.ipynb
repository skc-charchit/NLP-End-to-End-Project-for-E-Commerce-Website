## 1. Greedy vs Lazy Matching

- Greedy: Matches as much as possible

- Lazy: Matches as little as possible (add ? after quantifier)

```python
import re

text = "<tag>hello</tag><tag>world</tag>"

# Greedy (default)
print(re.findall(r'<tag>.*</tag>', text))  
# Output: ['<tag>hello</tag><tag>world</tag>']

# Lazy
print(re.findall(r'<tag>.*?</tag>', text))  
# Output: ['<tag>hello</tag>', '<tag>world</tag>']
```

## 2. Non-Capturing Groups

- Use (?:...) to group patterns without creating a capture group.

- Useful for applying quantifiers without storing the match.

```python
text = "apple apple apple"
pattern = r'(?:apple ){2}'  # matches "apple apple "
print(re.findall(pattern, text))
# Output: ['apple apple ']
```

## 3. Named Groups

- Capture groups with names instead of numbers.

- Makes extracting matches easier and readable.

```python
text = "Name: John, Age: 30"
match = re.search(r"Name: (?P<name>\w+), Age: (?P<age>\d+)", text)
print(match.group('name'))  # Output: John
print(match.group('age'))   # Output: 30
```

## 4. Lookahead & Lookbehind

- Positive Lookahead: (?=...) → matches only if followed by pattern

- Negative Lookahead: (?!...) → matches only if NOT followed by pattern

- Positive Lookbehind: (?<=...) → matches only if preceded by pattern

- Negative Lookbehind: (?<!...) → matches only if NOT preceded by pattern

```python
text = "10kg, 20lb, 30kg"

# Match numbers only followed by 'kg'
print(re.findall(r'\d+(?=kg)', text))  
# Output: ['10', '30']
```

## 5. Flags / Modifiers

Control regex behavior with flags:

---
| Flag            | Description                       |
| --------------- | --------------------------------- |
| `re.IGNORECASE` | Ignore case                       |
| `re.MULTILINE`  | `^` and `$` match line boundaries |
| `re.DOTALL`     | `.` matches newline too           |

---

```python
text = "Hello\nWorld"
print(re.findall(r'.+', text, re.DOTALL))
# Output: ['Hello\nWorld']
```

## 6. Unicode & Special Sequences

- Standard re supports \w, \d, etc., for ASCII

- For full Unicode properties (emojis, accented letters), use regex module:

```python
import regex
text = "I love Python 🐍 and Café"
print(regex.findall(r'\p{L}+', text))  
# Output: ['I', 'love', 'Python', 'and', 'Café']
```

## 7. Performance Tips

- Avoid catastrophic backtracking: nested ambiguous quantifiers.

- Split complex patterns into smaller steps for efficiency.

- Prefer non-capturing groups if you don’t need the match.

- Use re.compile() for patterns used multiple times.

```python
pattern = re.compile(r'\d+')
print(pattern.findall("My numbers are 12, 34, 56"))
```

## 8. Integration with Pandas

- Use regex to clean, extract, or transform text in datasets.

```python
import pandas as pd

df = pd.DataFrame({"text": ["Email: test@example.com", "Phone: 9876543210"]})

# Extract email
df['email'] = df['text'].str.findall(r'\S+@\S+\.\S+')

# Extract phone numbers
df['phone'] = df['text'].str.findall(r'\d+')

print(df)
```
