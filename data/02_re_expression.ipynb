# 📌 2. Regex Patterns (Core Types)

## A. Character Types

---
| Symbol | Meaning                               |
| ------ | ------------------------------------- |
| `.`    | Any character except newline          |
| `\d`   | Digit (0-9)                           |
| `\D`   | Not a digit                           |
| `\w`   | Word character (letters, digits, `_`) |
| `\W`   | Not a word character                  |
| `\s`   | Whitespace (space, tab, newline)      |
| `\S`   | Non-whitespace                        |

---

```python
import re
re.findall(r'\w+', "Hello, World! 123")  
# ['Hello', 'World', '123']
```

## B. Anchors

---
| Symbol | Meaning             |
| ------ | ------------------- |
| `^`    | Start of string     |
| `$`    | End of string       |
| `\b`   | Word boundary       |
| `\B`   | Not a word boundary |

---

```python
re.findall(r'^Hello', "Hello World")  # ['Hello']
re.findall(r'World$', "Hello World")  # ['World']
```

## C. Quantifiers

---
| Symbol  | Meaning                     |
| ------- | --------------------------- |
| `*`     | 0 or more repetitions       |
| `+`     | 1 or more repetitions       |
| `?`     | 0 or 1 repetition           |
| `{m}`   | Exactly m repetitions       |
| `{m,n}` | Between m and n repetitions |

---

```python
re.findall(r'\d{2,4}', "My pin 12345 and 678")  
# ['1234', '678']
```

## D. Sets & Ranges


- [abc] → match any one character a, b, or c

- [a-z] → lowercase letters

- [^0-9] → any character except digits


```python
re.findall(r'[aeiou]', 'hello world')  # ['e', 'o', 'o']
```

## E. Groups & Capturing

- () → capture a group

- | → OR operator

```python
text = "cat bat rat"
re.findall(r'(cat|rat)', text)  # ['cat', 'rat']
```

## F. Lookahead & Lookbehind (Advanced)

- (?=...) → Positive lookahead

- (?<=...) → Positive lookbehind

```python
text = "foo123 bar456"
re.findall(r'\d+(?= bar)', text)  # ['123']
```
