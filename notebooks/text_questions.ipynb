## 🔹 Section 1: Regex Extraction (15 exercises)

15. Extract **sentences ending with ! or ?**.

## 🔹 Section 2: Pandas Cleaning (15 exercises)

1. Remove **all digits** but keep letters.  
2. Remove **all special characters except @ and #**.  
3. Replace multiple `!` or `?` with a single one.  
4. Normalize **extra spaces** between words.  
5. Standardize all text to **Title Case**.  
6. Create a new column counting **URLs per row**.  
7. Create a new column counting **emojis per row**.  
8. Replace **all emails** with `[EMAIL]`.  
9. Replace **all phone numbers** with `[PHONE]`.  
10. Replace **all digits** with `0`.  
11. Detect and flag rows containing **profanity words**.  
12. Drop rows where text length is **less than 5 characters**.  
13. Keep only alphabetic tokens and drop the rest.  
14. Replace **repeated characters**: e.g., `"heeeellooo"` → `"hello"`.  
15. Remove words longer than 20 characters (noise/spam).
