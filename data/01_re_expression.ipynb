# 📌 1. Understand the Basics
- What is Regex?

- Regex is a pattern-matching language used to find, extract, or manipulate text.

- Python provides the re module for regex.

---
| Function       | Description                                |
| -------------- | ------------------------------------------ |
| `re.match()`   | Matches pattern at the **start** of string |
| `re.search()`  | Finds pattern **anywhere** in string       |
| `re.findall()` | Returns **all matches** as a list          |
| `re.sub()`     | **Substitute** text using a pattern        |
| `re.split()`   | **Split** string using a pattern           |

---

```python
import re

text = "My number is 12345 and pin 6789"

# Find all numbers
numbers = re.findall(r'\d+', text)
print(numbers)  # ['12345', '6789']

# Replace numbers with XXX
masked = re.sub(r'\d+', 'XXX', text)
print(masked)  # 'My number is XXX and pin XXX'
```
