# 📌 3. Manipulating Data with Regex

```python
import re

# A. Cleaning text
text = "Hello!!! 😍 Welcome ### 123"
clean = re.sub(r'[^\w\s]', '', text)  # remove punctuation and emojis
print(clean)  # 'Hello  Welcome 123'
```

```python
# B. Extracting patterns

emails = "My email is test@example.com and yours is hello@world.org"
found = re.findall(r'\S+@\S+\.\S+', emails)
print(found)  # ['test@example.com', 'hello@world.org']
```

```python
# C. Splitting text

text = "apple,banana;orange|grape"
fruits = re.split(r'[;,|]', text)
print(fruits)  # ['apple', 'banana', 'orange', 'grape']
```

```python
# D. Masking sensitive data

text = "My phone is 9876543210"
masked = re.sub(r'\d{10}', 'XXXXXXXXXX', text)
print(masked)  # 'My phone is XXXXXXXXXX'
```

# 📌 5. Tips to Master Regex

- Start small – test regex patterns in a Python shell.

- Use re.findall() to see all matches immediately.

- Visualize regex: https://regex101.com

- Combine regex + pandas to manipulate datasets efficiently:

```python
import pandas as pd

df = pd.DataFrame({"text":["Email: test@example.com", "Phone: 9876543210"]})
df['email'] = df['text'].str.findall(r'\S+@\S+\.\S+')
df['phone'] = df['text'].str.findall(r'\d+')
```
