# 📌 3. Manipulating Data with Regex

```python
import re

# A. Cleaning text
text = "Hello!!! 😍 Welcome ### 123"
clean = re.sub(r'[^\w\s]', '', text)  # remove punctuation and emojis
print(clean)  # 'Hello  Welcome 123'
```

```python
# B. Extracting patterns

emails = "My email is test@example.com and yours is hello@world.org"
found = re.findall(r'\S+@\S+\.\S+', emails)
print(found)  # ['test@example.com', 'hello@world.org']
```

```python
# C. Splitting text

text = "apple,banana;orange|grape"
fruits = re.split(r'[;,|]', text)
print(fruits)  # ['apple', 'banana', 'orange', 'grape']
```

```python
# D. Masking sensitive data

text = "My phone is 9876543210"
masked = re.sub(r'\d{10}', 'XXXXXXXXXX', text)
print(masked)  # 'My phone is XXXXXXXXXX'
```

## Regex for Emails, URLs, and Meaningful Info Extraction

```python
# 1. Removing Email IDs

import re

text = "Contact me at test@example.com or support@company.org"

# Remove emails
cleaned = re.sub(r'\b[\w.%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', '', text)
print(cleaned)
# Output: "Contact me at  or "
```

## 📌 Pattern explanation:

- [\w.%+-]+ → username part (letters, digits, ., %, +, -)

- @ → literal @

- [A-Za-z0-9.-]+ → domain

- \.[A-Z|a-z]{2,} → top-level domain like .com, .org

```python
# 2. Extracting Email IDs
emails = re.findall(r'\b[\w.%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text)
print(emails)
# Output: ['test@example.com', 'support@company.org']
```

```python
# 3. Removing URLs

text = "Check out https://www.example.com or http://test.org/page?id=5"

# Remove URLs
cleaned = re.sub(r'http\S+|www\S+', '', text)
print(cleaned)
# Output: "Check out  or "
```

## 📌 Pattern explanation:

- http\S+ → http followed by non-space characters

- www\S+ → www followed by non-space characters

```python
# 4. Extracting URLs
urls = re.findall(r'http\S+|www\S+', text)
print(urls)
# Output: ['https://www.example.com', 'http://test.org/page?id=5']
```

## 5. Extracting Meaningful Info from URLs

- Sometimes you don’t just want the URL, but components like domain, path, or query parameters.

```python
# Extract Domain
text = "Visit https://www.example.com/page?id=5"

domain = re.findall(r'https?://([^/]+)', text)
print(domain)
# Output: ['www.example.com']

```

```python
# Extract Path

path = re.findall(r'https?://[^/]+(/[^? ]*)', text)
print(path)
# Output: ['/page']
```

```python
# Extract Query Parameters

query = re.findall(r'\?(.*)', text)
print(query)
# Output: ['id=5']
```

## 6. Extracting Hashtags, Mentions, Numbers

```python
tweet = "Loving #Python 🤖! Follow @OpenAI for more updates. Version 3.10 released."

hashtags = re.findall(r'#\w+', tweet)
mentions = re.findall(r'@\w+', tweet)
numbers  = re.findall(r'\d+(\.\d+)?', tweet)

print(hashtags)  # ['#Python']
print(mentions)  # ['@OpenAI']
print(numbers)   # ['.10']
```

## 7. Remove Unnecessary Symbols but Keep Text

```python
text = "Hello!!! $$$ This is --bigcowboygruasfg-- with junk ###"

clean = re.sub(r'[^a-zA-Z0-9\s]', '', text)
print(clean)
# Output: "Hello  This is bigcowboygruasfg with junk "
```

```python
import re

# -------------------------
# Email Functions
# -------------------------

def remove_emails(text: str) -> str:
    """Remove all email addresses from text"""
    return re.sub(r'\b[\w.%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', '', text)

def extract_emails(text: str) -> list:
    """Extract all email addresses from text"""
    return re.findall(r'\b[\w.%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text)


# -------------------------
# URL Functions
# -------------------------

def remove_urls(text: str) -> str:
    """Remove all URLs from text"""
    return re.sub(r'http\S+|www\S+', '', text)

def extract_urls(text: str) -> list:
    """Extract all URLs from text"""
    return re.findall(r'http\S+|www\S+', text)

def extract_domains(text: str) -> list:
    """Extract only domain names from URLs in text"""
    return re.findall(r'https?://([^/]+)', text)

def extract_paths(text: str) -> list:
    """Extract URL paths (after domain)"""
    return re.findall(r'https?://[^/]+(/[^? ]*)', text)

def extract_queries(text: str) -> list:
    """Extract query parameters from URLs"""
    return re.findall(r'\?(.*)', text)


# -------------------------
# Social Media Functions
# -------------------------

def extract_hashtags(text: str) -> list:
    """Extract hashtags like #Python"""
    return re.findall(r'#\w+', text)

def extract_mentions(text: str) -> list:
    """Extract mentions like @user"""
    return re.findall(r'@\w+', text)


# -------------------------
# Numbers & Symbols
# -------------------------

def extract_numbers(text: str) -> list:
    """Extract integers and decimal numbers"""
    return re.findall(r'\d+(\.\d+)?', text)

def remove_symbols(text: str) -> str:
    """Remove special characters, keep alphanumeric + spaces"""
    return re.sub(r'[^a-zA-Z0-9\s]', '', text)


# -------------------------
# General Cleaner
# -------------------------

def clean_text(text: str) -> str:
    """Pipeline: remove emails, urls, and unnecessary symbols"""
    text = remove_emails(text)
    text = remove_urls(text)
    text = remove_symbols(text)
    return re.sub(r'\s+', ' ', text).strip()

```

## 📌 Usage Example

```python
sample = "Hello! Contact me at test@example.com 😎. Visit https://example.org/page?id=5 #Python @OpenAI"

print("Original:", sample)
print("Removed Emails:", remove_emails(sample))
print("Extracted Emails:", extract_emails(sample))
print("Removed URLs:", remove_urls(sample))
print("Extracted URLs:", extract_urls(sample))
print("Extracted Domain:", extract_domains(sample))
print("Extracted Path:", extract_paths(sample))
print("Extracted Query:", extract_queries(sample))
print("Hashtags:", extract_hashtags(sample))
print("Mentions:", extract_mentions(sample))
print("Numbers:", extract_numbers(sample))
print("Fully Cleaned:", clean_text(sample))
```
